```python
# import statment
import pandas as pd
import json
import numpy as np
import sqlite3
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
import matplotlib.pyplot as plt
import seaborn as sns
from ydata_profiling import ProfileReport
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import LabelEncoder
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
from scipy.stats import zscore
from scipy.stats.mstats import winsorize
```

```python
df = pd.read_csv(r"D:\RW\RW_Exam\Data_Preprocecing\Exam\Data\riders - riders.csv.csv")

with open("Data/trips.json", "r") as file:
    metadata = json.load(file)

conn = sqlite3.connect("city_zones.db")
with open(r"D:\RW\RW_Exam\Data_Preprocecing\Exam\Data\city_zones.sql", "r", encoding="utf-8") as file:
    sql_script = file.read()
conn.executescript(sql_script)
zones_df = pd.read_sql("SELECT * FROM city_zones", conn)

display(zones_df.head())
print(zones_df.info())
print(zones_df.isnull().sum())
print(zones_df.describe())


metadata_df = pd.DataFrame(metadata)
display(metadata_df.head())
print(metadata_df.info())
print(metadata_df.isnull().sum())
print(metadata_df.describe())

display(df.head())
print(df.info())
print(df.isnull().sum())
print(df.describe())

print("Duplicate rows:", df.duplicated().sum())
print("Invalid age entries:", df[df['age'] < 10].shape[0])
```

2

```python
#1. Handle numeric missing values using SimpleImputer (mean)
numeric_cols = metadata_df.select_dtypes(include='number').columns
num_imputer = SimpleImputer(strategy='mean')
metadata_df[numeric_cols] = num_imputer.fit_transform(metadata_df[numeric_cols])

#2. Handle categorical missing values using SimpleImputer (most frequent)
categorical_cols = metadata_df.select_dtypes(include='object').columns
cat_imputer = SimpleImputer(strategy='most_frequent')
metadata_df[categorical_cols] = cat_imputer.fit_transform(metadata_df[categorical_cols])

#3.Use KNN Imputer for multivariate columns.
knn_cols = ['duration_min', 'distance_km', 'fare_amount']
knn_imputer = KNNImputer(n_neighbors=5)
metadata_df[knn_cols] = knn_imputer.fit_transform(metadata_df[knn_cols])
```

```python
# 4. Convert date columns to datetime format
metadata_df['ride_date'] = pd.to_datetime(metadata_df['ride_date'])
df['signup_date'] = pd.to_datetime(df['signup_date'])
```

```python
# 5.Remove Unrealistic values
metadata_df = metadata_df[metadata_df['fare_amount'] >= 0]

metadata_df = metadata_df[
    ~((metadata_df['distance_km'] == 0) & 
      (metadata_df['fare_amount'] > 0))
]
```

3.

```python

before = metadata_df[['fare_amount', 'distance_km', 'duration_min']].agg(['mean', 'std', 'min', 'max'])

print("BEFORE:")
print(before)
```

```python
# Z-score detection
fare_z = zscore(metadata_df['fare_amount'])
distance_z = zscore(metadata_df['distance_km'])

fare_anomalies = metadata_df[abs(fare_z) > 3]
distance_anomalies = metadata_df[abs(distance_z) > 3]

print("Fare anomalies:", len(fare_anomalies))
print("Distance anomalies:", len(distance_anomalies))


# IQR for duration
Q1 = metadata_df['duration_min'].quantile(0.25)
Q3 = metadata_df['duration_min'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

duration_anomalies = metadata_df[
    (metadata_df['duration_min'] < lower) |
    (metadata_df['duration_min'] > upper)
]

print("Duration anomalies:", len(duration_anomalies))


# Winsorization
metadata_df['fare_amount'] = winsorize(
    metadata_df['fare_amount'],
    limits=[0.01, 0.01]
)
```

```python
after = metadata_df[['fare_amount', 'distance_km', 'duration_min']].agg(['mean', 'std', 'min', 'max'])

print("\nAFTER:")
print(after)

comparison = pd.concat([before, after],keys=['Before', 'After'])

print("\nBEFORE VS AFTER:")
print(comparison)
```

4

```python
# 1. Convert datetime → hour, day_of_week, month

metadata_df['ride_date'] = pd.to_datetime(metadata_df['ride_date'])

metadata_df['hour'] = metadata_df['ride_date'].dt.hour
metadata_df['day_of_week'] = metadata_df['ride_date'].dt.dayofweek
metadata_df['month'] = metadata_df['ride_date'].dt.month


# 2. Encode categorical columns
label_encoder = LabelEncoder()

if df['gender'].dtype == 'object':
    df['gender'] = label_encoder.fit_transform(df['gender'])


# Merge zone data

metadata_df = metadata_df.merge(zones_df,left_on='zone',right_on='zone_name',how='left')


# 3. Create traffic level from traffic index

metadata_df['traffic_level'] = pd.qcut(metadata_df['traffic_index'].rank(method='first'),q=3,labels=['Low', 'Medium', 'High'])


# 4. Ordinal Encode traffic level
ordinal_encoder = OrdinalEncoder(categories=[['Low', 'Medium', 'High']])
metadata_df['traffic_level'] = ordinal_encoder.fit_transform(metadata_df[['traffic_level']])


# 5. One-Hot Encode payment mode and zone
metadata_df = pd.get_dummies(metadata_df,columns=['payment_mode', 'zone'],drop_first=False)

# 6. Binning
df['ride_frequency'] = pd.qcut(df['total_rides'],q=3,labels=['Low', 'Medium', 'High'])

# 7. Transform skewed numeri columns.
metadata_df['fare_log'] = np.log1p(metadata_df['fare_amount'])
metadata_df['distance_log'] = np.log1p(metadata_df['distance_km'])

metadata_df['duration_sqrt'] = np.sqrt(metadata_df['duration_min'])

# Check final result

display(metadata_df.head())
print(metadata_df.shape)
```

5

```python
scale_cols = [
    'distance_km',
    'duration_min',
    'fare_amount',
    'population_density',
    'traffic_index',
    'avg_speed_kmph'
]

# Before scaling
before_scaling = metadata_df[scale_cols].agg(
    ['mean', 'std', 'min', 'max']
)

print("BEFORE SCALING")
display(before_scaling)


# StandardScaler
standard_scaler = StandardScaler()

standard_scaled = standard_scaler.fit_transform(
    metadata_df[scale_cols]
)

standard_df = pd.DataFrame(
    standard_scaled,
    columns=scale_cols,
    index=metadata_df.index
)

print("\nAFTER STANDARD SCALER")
display(
    standard_df.agg(['mean', 'std', 'min', 'max'])
)


# MinMaxScaler
minmax_scaler = MinMaxScaler()

minmax_scaled = minmax_scaler.fit_transform(
    metadata_df[scale_cols]
)

minmax_df = pd.DataFrame(
    minmax_scaled,
    columns=scale_cols,
    index=metadata_df.index
)

print("\nAFTER MINMAX SCALER")
display(
    minmax_df.agg(['mean', 'std', 'min', 'max'])
)
```

```python
# 6. Feature Construction

# avg_ride_distance
ride_distance = metadata_df.groupby('rider_id')['distance_km'].sum()

df['avg_ride_distance'] = (
    df['rider_id'].map(ride_distance) / df['total_rides'])


# avg_ride_fare
ride_fare = metadata_df.groupby('rider_id')['fare_amount'].sum()

df['avg_ride_fare'] = (
    df['rider_id'].map(ride_fare) / df['total_rides'])


# is_peak_hour
metadata_df['is_peak_hour'] = (
    metadata_df['hour'].isin([7, 8, 9, 18, 19, 20, 21])).astype(int)


# days_since_signup
df['signup_date'] = pd.to_datetime(df['signup_date'])

df['days_since_signup'] = (
    pd.Timestamp.today() - df['signup_date']).dt.days


# ride_cancellation_rate
df['ride_cancellation_rate'] = (
    df['cancelled_rides'] / df['total_rides'])


# surge_flag
metadata_df['surge_flag'] = ((metadata_df['fare_amount']/metadata_df['distance_km'].replace(0, np.nan))>(metadata_df['fare_amount']/metadata_df['distance_km'].replace(0, np.nan)).quantile(0.95)).astype(int)
```

```python
# 7. Final Dataset

# Merge cleaned & enriched datasets
final_df = metadata_df.merge(df,on='rider_id',how='left')

# Summary table
summary = pd.DataFrame({
    'Metric': [
        'Rows',
        'Missing Values',
        'Engineered Features'],
    'Before': [
        len(metadata_df),
        metadata_df.isnull().sum().sum(),
        0],
    'After': [
        len(final_df),
        final_df.isnull().sum().sum(),
        len(final_df.columns) - len(metadata_df.columns)]})

print(summary)


# Display final dataset
display(final_df.head())
print(final_df.shape)


# Export final dataset
final_df.to_csv(
    'final_prepared_rides_dataset.csv',
    index=False
)

print("Final dataset exported successfully.")
```
