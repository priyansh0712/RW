````python
from sklearn.impute import SimpleImputer
import numpy as np
from sklearn.impute import KNNImputer
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

#Mean
mean_imputer = SimpleImputer(strategy="mean")
df_mean = df.copy()
df_mean[["age","annual_income"]] = mean_imputer.fit_transform(
    df_mean[["age","annual_income"]]
)

#Median
median_imputer = SimpleImputer(strategy="median")
df_median = df.copy()
df_median[["loan_amount","credit_score"]] = median_imputer.fit_transform(
    df_median[["loan_amount","credit_score"]]
)





#Random Sample Imputation
df_random = df.copy()
df_random["annual_income_missing"] = df_random["annual_income"].isnull().astype(int)

random_sample = df_random["annual_income"].dropna()

missing_index = df_random[df_random["annual_income"].isnull()].index

df_random.loc[missing_index, "annual_income"] = np.random.choice(random_sample,size=len(missing_index),replace=True)


#KNN Imputation
df_knn = df.copy()
knn = KNNImputer(n_neighbors=5)

cols = ["age","annual_income","loan_amount","credit_score"]

df_knn[cols] = knn.fit_transform(df_knn[cols])


#Mice Imputation
df_mice = df.copy()
mice = IterativeImputer(random_state=42)

cols = ["age","annual_income","loan_amount","credit_score"]

df_mice[cols] = mice.fit_transform(df_mice[cols])

#Most Frequent
df_mice["gender"] = df_mice["gender"].fillna(df_mice["gender"].mode()[0])

df_mice["employment_type"] = df_mice["employment_type"].fillna(df_mice["employment_type"].mode()[0])

print(df_mice.isnull().sum())

````

# Part - D Outlier Handling

````python
from scipy import stats

z_scores = stats.zscore(df_mice["annual_income"])
df_mice = df_mice[abs(z_scores) < 3]
display("Z-score", df_mice.head())

# IQR method
Q1 = df_mice["loan_amount"].quantile(0.25)
Q3 = df_mice["loan_amount"].quantile(0.75)
IQR = Q3 - Q1
df_mice = df_mice[(df_mice["loan_amount"] >= Q1 - 1.5 * IQR) & (df_mice["loan_amount"] <= Q3 + 1.5 * IQR)]
display("IQR", df_mice.head())

# Percentile method
lower_percentile = df_mice["loan_amount"].quantile(0.01)
upper_percentile = df_mice["loan_amount"].quantile(0.99)
df_mice = df_mice[(df_mice["loan_amount"] >= lower_percentile) & (df_mice["loan_amount"] <= upper_percentile)]
display("Percentile", df_mice.head())

# Winsorization method
from scipy.stats.mstats import winsorize

df_mice["loan_amount"] = winsorize(df_mice["loan_amount"], limits=[0.01, 0.01])
display("Winsorization", df_mice.head())
````

# Part - E

````python
df["join_date"] = pd.to_datetime(df["join_date"])

df["join_year"] = df["join_date"].dt.year
df["join_month"] = df["join_date"].dt.month
df["join_day"] = df["join_date"].dt.day
df["join_weekday"] = df["join_date"].dt.day_name()

df.head()
````

### Encoding Categorical Variables

````python
from sklearn.preprocessing import OrdinalEncoder,OneHotEncoder

# 1.Ordinal Encoding
encoder = OrdinalEncoder(categories=[["Primary","Secondary","Graduate","Post-Graduate"]])
df_mice["education_encoded"] = encoder.fit_transform(df_mice[["education_level"]])
display("Ordinal",df_mice[["education_level", "education_encoded"]].head())

# 2. Label Encoding 
df["high_credit"] = (df_mice["credit_score"] > 700).astype(int)
display("Label",df_mice[["credit_score", "high_credit"]].head())

# 3. One-Hot Encoding
ohe = OneHotEncoder(drop="first",sparse_output=False,handle_unknown="ignore")
encoded = ohe.fit_transform(df[["region", "loan_purpose"]])
encoded_df = pd.DataFrame(encoded,columns=ohe.get_feature_names_out(["region", "loan_purpose"]),index=df.index)
df = pd.concat([df, encoded_df], axis=1)
df.head()
````

### Encoding Numerical Variables

````python
from sklearn.preprocessing import Binarizer
from sklearn.cluster import KMeans

# 1. Binning
df_mice["income_group"] = pd.cut(df_mice["annual_income"],bins=[0, 300000, 600000, 1000000, np.inf],labels=["Low","Medium","High","Very High"])
display("Binned", df_mice[["annual_income", "income_group"]].head(10))

# 2. Binerization
binarizer = Binarizer(threshold=700)
df_mice["credit_score_binary"] = binarizer.fit_transform(df_mice[["credit_score"]])

display("Binerized", df_mice[["credit_score", "credit_score_binary"]].head()) 

# 3. Quntile Binning
df_mice["income_quantile"] = pd.qcut(df_mice["annual_income"],q=4,labels=["Q1-Low","Q2","Q3","Q4-High"])
display("Quntile Binned", df_mice[["annual_income", "income_quantile"]].head(10))

# 4. K-Means Binning
income_data = df_mice[["annual_income"]].dropna()
kmeans = KMeans(n_clusters=4,random_state=42,n_init=10)
income_data["income_cluster"] = kmeans.fit_predict(income_data[["annual_income"]])
df_mice.loc[income_data.index, "income_cluster"] = (income_data["income_cluster"])

display("K-Means Binned", df_mice[["annual_income", "income_cluster"]].head(10))
````

# Part - F

````python
from sklearn.preprocessing import StandardScaler, Normalizer, MinMaxScaler,MaxAbsScaler,RobustScaler
num_cols = [
    "age",
    "annual_income",
    "loan_amount",
    "credit_score",
    "repayment_history",
    "transaction_count",
    "spending_ratio"
]

# 1. Z-Score Scaling
standard_scaler = StandardScaler()
df_standard = df_mice.copy()
df_standard[num_cols] = standard_scaler.fit_transform(df_standard[num_cols])
display("Z-Score", df_standard[num_cols].head())

#2. Normalization
normalizer = Normalizer(norm="l2")
df_normalized = df_mice.copy()
df_normalized[num_cols] = normalizer.fit_transform(df_normalized[num_cols])
display("Normalization", df_normalized[num_cols].head())

#3.Min-Max Scaling

minmax_scaler = MinMaxScaler()
df_minmax = df_mice.copy()

df_minmax[num_cols] = minmax_scaler.fit_transform(df_minmax[num_cols])
display("Min-Max", df_minmax[num_cols].head())

#4. Max-Abs Scaling

maxabs_scaler = MaxAbsScaler()
df_maxabs = df_mice.copy()

df_maxabs[num_cols] = maxabs_scaler.fit_transform(df_maxabs[num_cols])
display("Max-Abs", df_maxabs[num_cols].head())

#5. Robust Scaling

robust_scaler = RobustScaler()
df_robust = df_mice.copy()

df_robust[num_cols] = robust_scaler.fit_transform(df_robust[num_cols])
display("Robust", df_robust[num_cols].head())
````

# Part - G

````python
from sklearn.preprocessing import FunctionTransformer,PowerTransformer,StandardScaler,MinMaxScaler
from sklearn.compose import ColumnTransformer

# Log Transformation
log_transformer = FunctionTransformer(np.log1p)

df_mice["income_log"] = log_transformer.transform(df_mice[["annual_income"]])

# Reciprocal Transformation
reciprocal_transformer = FunctionTransformer(lambda x: 1 / (x + 1e-9))

df_mice["loan_reciprocal"] = reciprocal_transformer.transform(df_mice[["loan_amount"]])

# Square Root Transformation
sqrt_transformer = FunctionTransformer(np.sqrt)

df_mice["transaction_sqrt"] = sqrt_transformer.transform(df_mice[["transaction_count"]])

# Box-Cox
boxcox = PowerTransformer(method="box-cox",standardize=False)

df_mice["income_boxcox"] = boxcox.fit_transform(df_mice[["annual_income"]]).ravel()

# Yeo-Johnson
yeojohnson = PowerTransformer(method="yeo-johnson",standardize=False)

df_mice["loan_yeojohnson"] = yeojohnson.fit_transform(
df_mice[["loan_amount"]]).ravel()


preprocessor = ColumnTransformer(transformers=[("income_scaling",StandardScaler(),["annual_income"]),
        ("loan_scaling",MinMaxScaler(),["loan_amount"]),
        ("transaction_log",FunctionTransformer(np.log1p),["transaction_count"])],
    remainder="passthrough")

transformed_data = preprocessor.fit_transform(df_mice)



df_mice["debt_to_income_ratio"] = (df_mice["loan_amount"] /df_mice["annual_income"])
df_mice["avg_monthly_transactions"] = (df_mice["transaction_count"] / 6)
df_mice["spending_to_income_ratio"] = (df_mice["spending_ratio"] / 100)


display(df_mice[["annual_income","income_log","income_boxcox","loan_amount","loan_reciprocal","loan_yeojohnson","transaction_count","transaction_sqrt","debt_to_income_ratio","avg_monthly_transactions","spending_to_income_ratio"]].head())
````
