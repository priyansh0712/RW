# Q.1 Planning a Data Science Project

## Dataset

**Titanic Dataset**

## Project Goal

Predict whether a passenger survived or not based on Age, Gender, Fare,
and Passenger Class.

### Explanation

A Data Science project starts by selecting a dataset. After that, the
data is cleaned, analyzed, used to train a machine learning model,
evaluated, deployed, and monitored.

## Typical Workflow of a Data Science Project

### 1. Define the Problem

-   Identify the problem.
-   Decide what needs to be predicted.

### 2. Collect Data

-   Collect the Titanic dataset.

### 3. Data Preprocessing & Cleaning

-   Remove missing values.
-   Remove duplicates.
-   Handle outliers.

### 4. Exploratory Data Analysis (EDA)

-   Analyze data using statistics and graphs.

### 5. Feature Engineering

-   Select useful features.
-   Convert categorical values into numerical values.

### 6. Model Building

-   Train a Machine Learning model.

### 7. Model Evaluation

-   Check Accuracy, Precision, Recall, F1-Score.

### 8. Deployment

-   Deploy the model.

### 9. Monitoring & Maintenance

-   Update the model with new data.

``` python
pipeline = [
    "Define the Problem",
    "Collect Data",
    "Data Preprocessing & Cleaning",
    "Exploratory Data Analysis (EDA)",
    "Feature Engineering",
    "Model Building",
    "Model Evaluation",
    "Deployment",
    "Monitoring & Maintenance"
]

for i, step in enumerate(pipeline, 1):
    print(f"{i}. {step}")
```

------------------------------------------------------------------------

# Q.2 Framing a Machine Learning Problem

## Dataset

**Titanic Dataset**

## Features (Input)

-   Age
-   Gender
-   Fare
-   Passenger Class

## Target (Output)

Survived

-   0 = Not Survived
-   1 = Survived

## ML Problem Type

**Classification**

``` python
problem = {
    "Dataset": "Titanic",
    "Features": ["Age", "Gender", "Fare", "Passenger Class"],
    "Target": "Survived",
    "ML_Type": "Classification"
}

print(problem)
```

# Q.3 Tensors

## What is a Tensor?

A tensor is a multidimensional array used in Machine Learning and Deep
Learning.

  Tensor      Dimension
  ----------- -----------
  Scalar      0D
  Vector      1D
  Matrix      2D
  3D Tensor   3D

``` python
import torch

scalar = torch.tensor(10)
vector = torch.tensor([1,2,3])
matrix = torch.tensor([[1,2],[3,4]])
tensor3d = torch.tensor([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
])

print(scalar, scalar.shape, scalar.ndim)
print(vector, vector.shape, vector.ndim)
print(matrix, matrix.shape, matrix.ndim)
print(tensor3d, tensor3d.shape, tensor3d.ndim)
```

------------------------------------------------------------------------

# Q.4 Tensor Operations

## Generate Random Tensor

``` python
import torch

tensor = torch.rand(3,3,3)
print(tensor)
```

## Reshape

``` python
reshape_tensor = tensor.reshape(9,3)

```

## Slicing

``` python
print(tensor[:2])
```

## Matrix Multiplication

``` python
A = torch.tensor([[1,2],[3,4]])
B = torch.tensor([[5,6],[7,8]])

print(torch.matmul(A,B))
```

## Broadcasting

``` python
A = torch.tensor([[1],[2],[3]])

B = torch.tensor([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

print(A + B)
```
