```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
data = pd.read_csv('Titanic.csv')
data.head()
```

```python
print(data.isna().sum())
updated = data.drop('Age', axis=1)
updated = updated.drop('Cabin', axis=1)
updated.head()
```

```python
simple_imputer = SimpleImputer(strategy='most_frequent')
updated[['Embarked']] = simple_imputer.fit_transform(updated[['Embarked']])
print(updated.isna().sum())
```

```python
data = {
    "Age": [25, np.nan, 35, 40, np.nan],
    "Salary": [3000, 4000, np.nan, 5000, 6000]
}
df = pd.DataFrame(data)
simple_imputer = SimpleImputer(strategy='mean')
df[['Age', 'Salary']] = simple_imputer.fit_transform(df[['Age', 'Salary']])
print(df)

df2 = pd.DataFrame(data=data)
simple_imputer = SimpleImputer(strategy='median')
df2[['Age', 'Salary']] = simple_imputer.fit_transform(df2[['Age', 'Salary']])
print(df2)
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Gender": ["F", np.nan, "M", "M", np.nan],
    "City": ["New York", "Los Angeles", np.nan, "Chicago", "New York"]
}
df = pd.DataFrame(data)
print(df)
print("\n")
imputer_mode = SimpleImputer(strategy="most_frequent")
df_mode = df.copy()
df_mode[["Gender", "City"]] = imputer_mode.fit_transform(df[["Gender", "City"]])
print(df_mode)

print("\n")
imputer_const = SimpleImputer(strategy="constant", fill_value="Unknown")
df_const = df.copy()
df_const[["Gender", "City"]] = imputer_const.fit_transform(df[["Gender", "City"]])
print(df_const)
```
