```python
import pandas as pd
import numpy as np

data = {
    'Age':    [22,  np.nan, 25,  23,  np.nan, 21,  24,  np.nan],
    'Income': [30000, np.nan, 45000, np.nan, 28000, np.nan, 52000, np.nan],
}

data = pd.DataFrame(data)
```

```python
data["Age_missing"] = data["Age"].isnull().astype(int)
data["Income_missing"] = data["Income"].isnull().astype(int)
```

```python
data.head()
```

```python
random_sample = data['Age'].dropna().sample(data['Age'].isna().sum(), random_state=42)
random_sample.index = data[data['Age'].isna()].index
data.loc[data['Age'].isna(), 'Age'] = random_sample

print(data)
```

# KNN impute

```python
from sklearn.impute import KNNImputer
```

```python
data = {
    'Rider':           ['Raju', 'Suresh', 'Amit', 'Vikram', 'Pooja',
                        'Kiran', 'Deepak', 'Meena', 'Arjun', 'Sunita'],
    'Deliveries_Day':  [18,   25,      np.nan, 30,      15,
                        22,   28,      np.nan, 19,      26     ],
    'Exp_Years':       [3,    5,       4,      6,       np.nan,
                        4,    5,       3,      np.nan,  5      ],
    'Avg_Distance_km': [8.5,  12.0,    10.5,   np.nan,  7.0,
                        9.5,  np.nan,  8.0,    9.0,     11.5   ],
    'Rating':          [4.2,  4.7,     np.nan, 4.9,     4.0,
                        4.4,  4.8,     4.1,    np.nan,  4.6    ],
}
data = pd.DataFrame(data)
display(data)
```

```python
print(data.isnull().sum())
```

```python
KNN_imputer = KNNImputer(n_neighbors=2,weights='distance')
data_imputed = KNN_imputer.fit_transform(data[['Deliveries_Day', 'Exp_Years', 'Avg_Distance_km', 'Rating']])

data_imputed_df = pd.DataFrame(data_imputed, columns=['Deliveries_Day', 'Exp_Years', 'Avg_Distance_km', 'Rating'])
data_imputed_df.head()
```
