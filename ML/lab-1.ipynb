```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression 
```

```python
actual_price = np.random.randint(2000, 25550, size=10)
predicted_price = np.random.randint(2000, 25550, size=10)

mean_squared_error = np.mean((actual_price - predicted_price) ** 2)
print("Mean Squared Error:", mean_squared_error)

mean_absolute_error = np.mean(np.abs(actual_price - predicted_price))
print("Mean Absolute Error:", mean_absolute_error)

root_mean_squared_error = np.sqrt(mean_squared_error)
print("Root Mean Squared Error:", root_mean_squared_error)
```

```python
data = {
    'Experience': np.abs(np.random.randint(0, 10, size=10)),
    'Salary': np.random.randint(30000, 75000, size=10)
}

model = LinearRegression()
model.fit(np.array(data['Experience']).reshape(-1, 1), data['Salary'])

new_experience = np.array([[5], [7], [10], [12], [15]])
predicted_salary = model.predict(new_experience)

for exp, salary in zip(new_experience.flatten(), predicted_salary):
    print(f"Predicted salary for {exp} years of experience: ${salary:.2f}")
    
mse = np.mean((np.array(data['Salary']) - model.predict(np.array(data['Experience']).reshape(-1, 1))) ** 2)
print(f"Mean Squared Error: {mse:.2f}")

mae = np.mean(np.abs(np.array(data['Salary']) - model.predict(np.array(data['Experience']).reshape(-1, 1))))
print(f"Mean Absolute Error: {mae:.2f}")

rmse = np.sqrt(mse)
print(f"Root Mean Squared Error: {rmse:.2f}")
```
