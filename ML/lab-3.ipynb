```python
import numpy as np

X = np.array([
    [1, 2, 1],
    [1, 3, 2],
    [1, 4, 1],
    [1, 5, 2],
    [1, 6, 1],
    [1, 7, 2],
    [1, 8, 1],
    [1, 9, 2],
    [1, 10, 3],
    [1, 11, 3]
])

beta = np.array([
    [5000],
    [2000],
    [1500]
])

Y_pred = X @ beta

print("X shape:", X.shape)
print("Beta shape:", beta.shape)
print("Prediction shape:", Y_pred.shape)

print("\nPredicted Y:")
print(Y_pred)
```
