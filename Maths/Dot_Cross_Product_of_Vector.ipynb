# 🔹 Part 1: Distance Between Two Points

Euclidean Distance Formula:

Distance = √[(x₂ - x₁)² + (y₂ - y₁)²]

📘 Example: Customer location tracking


```python

import math
import numpy as np
import matplotlib.pyplot as plt

# Distance Between Two Points

# -----------------------------
# 2D Example
# -----------------------------
P1 = np.array([2, 3])
P2 = np.array([6, 7])

manual_distance_2d = math.sqrt(((P2 - P1)**2).sum())
distance = np.linalg.norm(P2 - P1)

# -----------------------------
# 3D Example
# -----------------------------
Q1 = np.array([1, 2, 3])
Q2 = np.array([4, 6, 8])

distance_3d = math.sqrt(((Q2 - Q1)**2).sum())

# -----------------------------
# Output
# -----------------------------
print("2D Points:", P1, P2)
print("Distance (2D):", manual_distance_2d)
print("Distance (2D):", distance)

print("\n3D Points:", Q1, Q2)
print("Distance (3D):", distance_3d)
```

```python

A = np.array([4, 3])
B = np.array([1, 2])

plt.scatter(A[0], A[1])
plt.scatter(B[0], B[1])
plt.plot([A[0], B[0]], [A[1], B[1]])
plt.title("Distance Between Two Points")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```


# 🔹 Part 2: Norms

Norms measure the length or size of a vector.

A norm is a function that assigns a positive length to a vector.

It tells: “How big is this vector?”

## L1 Norm (Manhattan Distance)
Also called:
- Taxicab distance
- Manhattan distance

||x||₁ = |x₁| + |x₂| + ...

image.png


## L2 Norm (Euclidean Norm)
||x||₂ = √(x₁² + x₂² + ...)

Used in Machine Learning regularization.

image.png

```python
# Norms (L1 & L2)

# Define vector
v = np.array([3, -4, 5])

# -----------------------------
# Manual L1 Norm
# -----------------------------
L1_manual = np.sum(np.abs(v))  # math.fabs()

# -----------------------------
# Manual L2 Norm
# -----------------------------
L2_manual = np.sqrt(np.sum(v**2))

# -----------------------------
# Using NumPy Built-in
# -----------------------------
L1_numpy = np.linalg.norm(v, ord=1)
L2_numpy = np.linalg.norm(v, ord=2)

# -----------------------------
# Output
# -----------------------------
print("Vector:", v)

print("\nL1 Norm (Manual):", L1_manual)
print("L1 Norm (NumPy):", L1_numpy)

print("\nL2 Norm (Manual):", L2_manual)
print("L2 Norm (NumPy):", L2_numpy)
```

# 🔹 Part 3: Unit Vector

A unit vector is a vector that has:

**Magnitude = 1**

It represents direction only, not size.

**Unit Vector = v / |v|**

image.png


```python
# Unit Vector

# Define vector
v = np.array([3, 4])

# -----------------------------
# Step 1: Compute Magnitude
# -----------------------------
magnitude = np.linalg.norm(v)

# -----------------------------
# Step 2: Compute Unit Vector
# -----------------------------
unit_vector_manual = v / magnitude

# Verify magnitude of unit vector
unit_magnitude = np.linalg.norm(unit_vector_manual)

# -----------------------------
# Output
# -----------------------------
print("Original Vector:", v)
print("Magnitude:", magnitude)

print("\nUnit Vector:", unit_vector_manual)
print("Magnitude of Unit Vector:", unit_magnitude)
```


# 🔹 Part 4: Dot Product

The dot product of two vectors gives a single number (scalar).

It measures:

How much two vectors point in the same direction.


A · B = x₁y₁ + x₂y₂



image.png

image.png

```python
# Dot Product

# Define vectors
a = np.array([2, 3])
b = np.array([4, 1])

# -----------------------------
# 1. Manual Dot Product
# -----------------------------
dot_manual = a[0]*b[0] + a[1]*b[1]

# -----------------------------
# 2. NumPy Dot Product
# -----------------------------
dot_numpy = np.dot(a, b)

# -----------------------------
# 3. Magnitudes
# -----------------------------
mag_a = np.linalg.norm(a)
mag_b = np.linalg.norm(b)

# -----------------------------
# 4. Angle Between Vectors
# -----------------------------
cos_theta = dot_numpy / (mag_a * mag_b)
theta = math.degrees(math.acos(cos_theta))

# -----------------------------
# Output
# -----------------------------
print("Vector a:", a)
print("Vector b:", b)

print("\nManual Dot Product:", dot_manual)
print("NumPy Dot Product:", dot_numpy)

print("\nMagnitude of a:", mag_a)
print("Magnitude of b:", mag_b)

print("\nAngle between vectors (degrees):", theta)
```

# 🔹 Part 5: Cross Product (3D)

The cross product of two vectors:
- Produces another vector
- That is perpendicular to both original vectors
- Defined only in 3D


image.png

image.png

```python
# Cross Product

# Define vectors
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# -----------------------------
# 1. Manual Cross Product
# -----------------------------
cross_manual = np.array([
    a[1]*b[2] - a[2]*b[1],
    a[2]*b[0] - a[0]*b[2],
    a[0]*b[1] - a[1]*b[0]
])

# -----------------------------
# 2. NumPy Cross Product
# -----------------------------
cross_numpy = np.cross(a, b)

# -----------------------------
# 3. Verify Perpendicularity
# Dot product should be 0
# -----------------------------
dot_with_a = np.dot(cross_numpy, a)
dot_with_b = np.dot(cross_numpy, b)

# -----------------------------
# Output
# -----------------------------
print("Vector a:", a)
print("Vector b:", b)

print("\nManual Cross Product:", cross_manual)
print("NumPy Cross Product:", cross_numpy)

print("\nDot(cross, a):", dot_with_a)
print("Dot(cross, b):", dot_with_b)
```
