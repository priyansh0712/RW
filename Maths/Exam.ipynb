```python
import numpy as np

data = np.random.randint(0, 100,(50))
mu = 70
sigma = 8
x_bar = np.mean(data, axis=0)

# Z-score
z_scores = (x_bar - mu) / (sigma / np.sqrt(data.shape[0]))
print("Z-scores for each column:", z_scores)

# P value
p_values = 2 * (1 - np.abs(z_scores))
print("P-values for each column:", p_values)
```

```python
# generate male heigh and female height data assume both population have knon SD perform a two-sample z-test and print z-value and p-value
# Generate male height data
male_heights = np.random.normal(loc=175, scale=10, size=100)  # mean=175 cm, std=10 cm, n=100
# Generate female height data
female_heights = np.random.normal(loc=162, scale=10, size=100)  # mean=162 cm, std=10 cm, n=100
# Calculate means and standard deviations
mean_male = np.mean(male_heights)
mean_female = np.mean(female_heights)

sigma_male= 10
sigma_female = 10

z_score = (mean_male - mean_female) / np.sqrt((sigma_male**2 / len(male_heights)) + (sigma_female**2 / len(female_heights)))
p_value = 2 * (1 - np.abs(z_score))
print("Z-value for two-sample z-test:", z_score)
print("P-value for two-sample z-test:", p_value)
```

```python
water_intake = np.random.normal(loc=2, scale=0.5, size=40)  # mean=2 liters, std=0.5 liters, n=40
population_mean = 2
population_std = 0.4
sample_mean = np.mean(water_intake)
z_score_one_sample = (sample_mean - population_mean) / (population_std / np.sqrt(len(water_intake)))
p_value_one_sample = 2 * (1 - np.abs(z_score_one_sample))
print("Z-value for one-sample z-test:", z_score_one_sample)
print("P-value for one-sample z-test:", p_value_one_sample)
```

```python
#generate two dataset of 45 and 50 student score respectivly.
scores_class_a = np.random.normal(loc=85, scale=5, size=45)  # mean=85, std=5, n=45
scores_class_b = np.random.normal(loc=80, scale=5, size=50)

#Assume knonn SD for both population perform a two-sample z-test and print z-value and p-value
mean_a = np.mean(scores_class_a)
mean_b = np.mean(scores_class_b)
std_a = np.std(scores_class_a, ddof=1)
std_b = np.std(scores_class_b, ddof=1)
n_a = len(scores_class_a)
n_b = len(scores_class_b)
z_score_two_sample = (mean_a - mean_b) / np.sqrt((std_a**2)/n_a + (std_b**2)/n_b)
p_value_two_sample = 2 * (1 - np.abs(z_score_two_sample))
print("Z-value for two-sample z-test:", z_score_two_sample)
print("P-value for two-sample z-test:", p_value_two_sample)
```

```python
sleep_hours = np.random.normal(loc=7, scale=1.5, size=30)  # mean=7 hours, std=1.5 hours, n=30
population_mean_sleep = 8

sample_mean_sleep = np.mean(sleep_hours)
sample_std_sleep = np.std(sleep_hours, ddof=1)
t_statistic = (sample_mean_sleep - population_mean_sleep) / (sample_std_sleep / np.sqrt(len(sleep_hours)))
degrees_of_freedom = len(sleep_hours) - 1
from scipy import stats
p_value_t_test = 2 * (1 - stats.t.cdf(np.abs(t_statistic), df=degrees_of_freedom))
print("T-statistic for one-sample t-test:", t_statistic)
print("P-value for one-sample t-test:", p_value_t_test)
```

```python
# Generate two independend sample : 40values for smokers and 40 for non-smokers
smokers = np.random.normal(loc=10, scale=2, size=40)
non_smokers = np.random.normal(loc=8, scale=2, size=40)

#Assume equal or unequal variances as needed
mean_smokers = np.mean(smokers)
mean_non_smokers = np.mean(non_smokers)
std_smokers = np.std(smokers, ddof=1)
std_non_smokers = np.std(non_smokers, ddof=1)

# perform independent two-sample t-test
t_statistic_independent = (mean_smokers - mean_non_smokers) / np.sqrt((std_smokers**2 / len(smokers)) + (std_non_smokers**2 / len(non_smokers)))
degrees_of_freedom_independent = len(smokers) + len(non_smokers) - 2
p_value_independent = 2 * (1 - stats.t.cdf(np.abs(t_statistic_independent), df=degrees_of_freedom_independent))
print("T-statistic for independent two-sample t-test:", t_statistic_independent)
print("P-value for independent two-sample t-test:", p_value_independent)
```

```python
# generate a sample of 25 teenagers daily screen time (in hours).
screen_time = np.random.normal(loc=4, scale=1, size=25)  # mean=4 hours, std=1 hour, n=25
population_mean_screen_time = 4
sample_mean_screen_time = np.mean(screen_time)
sample_std_screen_time = np.std(screen_time, ddof=1)
t_statistic_screen_time = (sample_mean_screen_time - population_mean_screen_time) / (sample_std_screen_time / np.sqrt(len(screen_time)))
degrees_of_freedom_screen_time = len(screen_time) - 1
p_value_screen_time = 2 * (1 - stats.t.cdf(np.abs(t_statistic_screen_time), df=degrees_of_freedom_screen_time))
print("T-statistic for one-sample t-test:", t_statistic_screen_time)
print("P-value for one-sample t-test:", p_value_screen_time)
```
