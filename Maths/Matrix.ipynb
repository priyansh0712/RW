```python
import numpy as np

a = np.array([1,2])
b = np.array([3,4])

angle = np.arccos(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))
print(np.degrees(angle))

a = np.random.rand(3)
b = np.random.rand(3)

angle = np.arccos(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))
print(np.degrees(angle))
```

```python
import matplotlib.pyplot as plt
a = np.array([3,4])
b = np.array([1,0.5])
proj = (np.dot(a,b)/np.linalg.norm(b)**2) * b
print(proj)

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='r', label='a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='black', label='b')
plt.quiver(0, 0, proj[0], proj[1], angles='xy', scale_units='xy', scale=1, color='y', label='projection of a onto b')
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.show()
```

```python
x = np.linspace(0,10,100)
y = 2*x-1

plt.plot(x,y)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X,Y = np.meshgrid(x,y)
Z = 5-X-Y

fig=plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X,Y,Z)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot of Z = 5 - X - Y')
plt.show()
```
