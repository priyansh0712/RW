# 2. Explain the difference between

## Mean, Median, and Mode

| Measure | Definition |
|----------|------------|
| Mean | Arithmetic average of all values |
| Median | Middle value after sorting |
| Mode | Most frequently occurring value |

### Example

Dataset: 10, 20, 20, 30, 40

Mean:

$$
\frac{10+20+20+30+40}{5}=24
$$

Median = 20

Mode = 20

### Difference

- Mean considers every value.
- Median is resistant to outliers.
- Mode identifies the most frequent observation.

---

## Range, Variance, and Standard Deviation

| Measure | Definition |
|----------|------------|
| Range | Difference between maximum and minimum value |
| Variance | Average squared deviation from mean |
| Standard Deviation | Square root of variance |

### Formula for Range

$$
Range = Maximum - Minimum
$$

### Formula for Variance

$$
\sigma^2 = \frac{\sum (x-\mu)^2}{N}
$$

### Formula for Standard Deviation

$$
\sigma = \sqrt{\sigma^2}
$$

### Example

Dataset: 10, 20, 30

Range:

$$
30-10=20
$$

Variance:

$$
66.67
$$

Standard Deviation:

$$
8.16
$$

# 3.1 Gaussian Distribution (Normal Distribution)

## Definition

A Gaussian Distribution is a symmetric bell-shaped probability distribution in which:

$$
Mean = Median = Mode
$$

## Formula

$$
f(x)=\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Where:

- μ = Mean
- σ = Standard Deviation

## Diagram

```text
                 Mean (μ)
                    |
                    |
              .-'''''''-.
           .-'           '-.
         .'                 '.
        /                     \
-------/-----------------------\-------
```

## Characteristics

- Bell-shaped curve
- Symmetrical distribution
- Mean = Median = Mode
- Widely used in statistics

---

# 3.2 Log-Normal Distribution

## Definition

A variable follows a Log-Normal Distribution when its logarithm follows a Normal Distribution.

## Formula

$$
Y = \ln(X)
$$

Where:

- X = Original Variable
- Y = Logarithm of X

## Diagram

```text
        /\
       /  \
      /    \____
     /           \____
----/--------------------->
```

## Characteristics

- Positively skewed
- Long right tail
- Cannot contain negative values

## Real Life Example

- Income Distribution
- Stock Prices
- Property Prices

---

# 3.3 3-Sigma Rule (Empirical Rule)

## Definition

The Empirical Rule describes how data is distributed in a normal distribution.

## Formula

$$
\mu \pm 1\sigma
$$

$$
\mu \pm 2\sigma
$$

$$
\mu \pm 3\sigma
$$

## Diagram

```text
          68%       95%        99.7%

----|----|----|----|----|----|----
   -3σ  -2σ  -1σ   μ  +1σ +2σ +3σ
```

## Interpretation

- 68% of observations lie within ±1σ
- 95% of observations lie within ±2σ
- 99.7% of observations lie within ±3σ

---

# 3.4 Percentiles

## Definition

A percentile indicates the percentage of observations below a particular value.

## Formula

$$
P_k = \frac{k}{100}(n+1)
$$

Where:

- k = Percentile Number
- n = Number of Observations

## Diagram

```text
0%      25%      50%      75%      100%
|--------|--------|--------|---------|
 Min      P25     P50      P75      Max
```

## Example

90th Percentile means:

90% of observations are below that value.

---

# 3.5 Quartiles

## Definition

Quartiles divide data into four equal parts.

## Formula

$$
Q_1 = 25th\ Percentile
$$

$$
Q_2 = Median
$$

$$
Q_3 = 75th\ Percentile
$$

## Diagram

```text
|------|------|------|------|
Min    Q1     Q2     Q3    Max
```

## Interpretation

- Q1 contains first 25% data
- Q2 contains first 50% data
- Q3 contains first 75% data

---

# 3.6 Five Number Summary

## Definition

Five Number Summary is a set of descriptive statistics consisting of:

1. Minimum
2. Q1
3. Median
4. Q3
5. Maximum

## Diagram

```text
Min      Q1     Median     Q3      Max
 |--------|--------|--------|--------|
```

## Uses

- Boxplot Construction
- Outlier Detection
- Data Summary

---

# 3.7 Skewness

## Definition

Skewness measures the asymmetry of a distribution.

## Formula

$$
Skewness=
\frac{\sum (x_i-\bar{x})^3}
{(n-1)s^3}
$$

## Positive Skew

```text
      /\
     /  \
    /    \______
---/---------------> 
```

## Negative Skew

```text
 ______
/      \
        \
         \
----------\-------->
```

## Symmetric Distribution

```text
       /\
      /  \
     /    \
----/------\----
```

## Interpretation

- Positive Skew → Right Tail
- Negative Skew → Left Tail
- Zero Skew → Symmetric Distribution

---

# 3.8 Kurtosis

## Definition

Kurtosis measures the heaviness of tails and presence of outliers in a distribution.

## Formula

$$
Kurtosis=
\frac{\sum (x_i-\bar{x})^4}
{(n-1)s^4}
$$

## Leptokurtic (High Kurtosis)

```text
         /\
        /  \
       /    \
------/------\------
```

## Mesokurtic (Normal Kurtosis)

```text
      /\
     /  \
----/----\----
```

## Platykurtic (Low Kurtosis)

```text
    ______
---/      \---
```

## Interpretation

- Leptokurtic → Heavy Tails
- Mesokurtic → Normal Distribution
- Platykurtic → Light Tails

````python
print("Mean Age:",
      df['Age_of_Household_Head'].mean())
print("Median Age:",
      df['Age_of_Household_Head'].median())
print("Mode Age:",
      df['Age_of_Household_Head'].mode()[0])
````

````python
income_range = (df['Household_Income'].max()-df['Household_Income'].min())
variance_income = df['Household_Income'].var()
std = df['Household_Income'].std()
Q1 = df['Household_Income'].quantile(0.25)
Q3 = df['Household_Income'].quantile(0.75)
IQR = Q3 - Q1
print("Income Range:", income_range)
print("Variance of Income:", variance_income)
print("Standard Deviation of Income:", std)
print("First Quartile:", Q1)
print("Third Quartile:", Q3)
print("Interquartile Range:", IQR)
````

- Interpretation
    - Large IQR → High income spread
    - Small IQR → Low income spread
