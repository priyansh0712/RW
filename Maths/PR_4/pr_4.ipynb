1️⃣ Vector & Matrix Fundamentals

```python
import numpy as np
import matplotlib.pyplot as plt
# ===========================================
# Student Subject Score Vectors
# ===========================================

student1_scores = np.array([80, 70, 90])
student2_scores = np.array([60, 85, 75])

print("Student 1 Scores:", student1_scores)
print("Student 2 Scores:", student2_scores)

# ===========================================
# L1 Norm (Manhattan Norm)
# ===========================================

student1_l1_norm = np.linalg.norm(student1_scores, ord=1)
student2_l1_norm = np.linalg.norm(student2_scores, ord=1)

print("\n----- L1 Norm -----")
print("Student 1 L1 Norm:", student1_l1_norm)
print("Student 2 L1 Norm:", student2_l1_norm)

# ===========================================
# L2 Norm (Euclidean Norm)
# ===========================================

student1_l2_norm = np.linalg.norm(student1_scores)
student2_l2_norm = np.linalg.norm(student2_scores)

print("\n----- L2 Norm -----")
print("Student 1 L2 Norm:", round(student1_l2_norm, 2))
print("Student 2 L2 Norm:", round(student2_l2_norm, 2))

# ===========================================
# Dot Product
# ===========================================

dot_product = np.dot(student1_scores, student2_scores)

print("\n----- Dot Product -----")
print("Dot Product:", dot_product)

# ===========================================
# Angle Between Two Vectors
# ===========================================

cosine_value = dot_product / (student1_l2_norm * student2_l2_norm)

angle_in_radians = np.arccos(cosine_value)
angle_in_degrees = np.degrees(angle_in_radians)

print("\n----- Angle Between Vectors -----")
print("Angle (Radians):", round(angle_in_radians, 4))
print("Angle (Degrees):", round(angle_in_degrees, 2))

# ===========================================
# Cross Product
# ===========================================

cross_product = np.cross(student1_scores, student2_scores)

print("\n----- Cross Product -----")
print("Cross Product:", cross_product)

# ===========================================
# Projection of Student 1 onto Student 2
# ===========================================

projection_of_student1_on_student2 = (np.dot(student1_scores,student2_scores)/np.linalg.norm(student2_scores)**2)*student2_scores

print("\n----- Projection of Student 1 onto Student 2 -----")
print(np.round(projection_of_student1_on_student2, 2))

# ==========================================
# 2D Visualization
# (Using Math & Science Scores)
# ==========================================
plt.figure(figsize=(7,7))

plt.quiver(
    0, 0,
    student1_scores[0],
    student1_scores[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='blue',
    label='Student A'
)

plt.quiver(
    0, 0,
    student2_scores[0],
    student2_scores[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color='red',
    label='Student B'
)

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.grid(True)
plt.xlabel("Math Score")
plt.ylabel("Science Score")
plt.title("Student Score Vectors (2D)")
plt.legend()

plt.show()

# ==========================================
# 3D Visualization
# ==========================================
fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(111, projection='3d')

ax.quiver(
    0,0,0,
    student1_scores[0],
    student1_scores[1],
    student1_scores[2],
    color='blue',
    label='Student A'
)

ax.quiver(
    0,0,0,
    student2_scores[0],
    student2_scores[1],
    student2_scores[2],
    color='red',
    label='Student B'
)

ax.set_xlabel("Math")
ax.set_ylabel("Science")
ax.set_zlabel("English")

ax.set_title("Student Score Vectors (3D)")

plt.legend()
plt.show()
```

2️⃣ Matrix Operations

```python
import numpy as np
import matplotlib.pyplot as plt

student_subject_matrix = np.array([
    [80, 70, 90],   
    [60, 85, 75],   
    [95, 88, 92]    
])

print("Student Subject Matrix:\n")
print(student_subject_matrix)

# ===========================================
# Matrix Addition
# ===========================================

matrix_addition = student_subject_matrix + student_subject_matrix

print("\n========== Matrix Addition ==========")
print(matrix_addition)

# ===========================================
# Matrix Multiplication
# ===========================================

matrix_multiplication = np.dot(student_subject_matrix,student_subject_matrix.T)

print("\n========== Matrix Multiplication ==========")
print(matrix_multiplication)

# ===========================================
# Matrix Transpose
# ===========================================

matrix_transpose = student_subject_matrix.T

print("\n========== Matrix Transpose ==========")
print(matrix_transpose)

# ===========================================
# Square Matrix for Inverse & Determinant
# ===========================================

square_matrix = np.array([[80, 70, 90],[60, 85, 75],[95, 88, 92]])

# ===========================================
# Determinant
# ===========================================

matrix_determinant = np.linalg.det(square_matrix)

print("\n========== Determinant ==========")
print(round(matrix_determinant, 2))

# ===========================================
# Matrix Inverse (If Possible)
# ===========================================

if matrix_determinant != 0:
    matrix_inverse = np.linalg.inv(square_matrix)

    print("\n========== Matrix Inverse ==========")
    print(np.round(matrix_inverse, 3))
else:
    print("\nInverse does not exist because determinant is 0.")

# ===========================================
# Visualization
# ===========================================

plt.figure(figsize=(6, 5))

plt.imshow(student_subject_matrix)

plt.colorbar(label="Marks")

plt.xticks([0, 1, 2],["Math", "Science", "English"])

plt.yticks([0, 1, 2],["Student 1", "Student 2", "Student 3"])

plt.title("Student Subject Matrix")

# Display values inside cells
for row in range(student_subject_matrix.shape[0]):
    for column in range(student_subject_matrix.shape[1]):
        plt.text(column,row,student_subject_matrix[row, column],ha="center",va="center",color="white",fontsize=12)

plt.show()
```

# Part C: Theory

## 1. Line

A **line** is a one-dimensional geometric object. In this dataset, if we consider only **two subjects**, such as **Math** and **Science**, each student is represented as a point in a 2D coordinate system. The relationship between these two features can be visualized on a plane using a line graph or scatter plot.

**Example:**

- Student 1 → (80, 70)
- Student 2 → (60, 85)
- Student 3 → (95, 88)

This represents the dataset in **2 dimensions (2D)**.

---
## 2. Plane

A **plane** is a two-dimensional flat surface that exists in **three-dimensional space**. When the dataset contains **three subjects**, such as **Math, Science, and English**, each student is represented by three coordinates.

**Example:**

- Student 1 → (80, 70, 90)
- Student 2 → (60, 85, 75)
- Student 3 → (95, 88, 92)

These points are visualized in **3D space**, where the three axes represent the three subjects.

---

## 3. Hyperplane

A **hyperplane** is a generalization of a line and a plane to higher dimensions. When the dataset contains **four or more features**, it cannot be visualized directly in 3D.

For example, if the dataset contains:

- Math
- Science
- English
- Computer
- Statistics

then each student is represented by **five coordinates**, forming a **5-dimensional dataset**. Such datasets are represented using **hyperplanes** in higher-dimensional space.

Hyperplanes are widely used in machine learning algorithms such as **Support Vector Machine (SVM)** and **Linear Discriminant Analysis (LDA)** to separate different classes of data.

---

# Dimensionality Increase

The dimensionality of the dataset increases as more subjects (features) are added.

| Number of Features | Representation |
|-------------------|----------------|
| 2 Features | Line (2D Space) |
| 3 Features | Plane (3D Space) |
| 4 Features | Hyperplane (4D Space) |
| 5 or More Features | Higher-Dimensional Hyperplane |

---

## Interpretation

- Using **2 subjects**, the student dataset is represented in **2D**.
- Using **3 subjects**, the dataset is represented in **3D**.
- When additional subjects are included, the dataset moves into **higher-dimensional space**, where a **hyperplane** is used to represent and analyze the data.
- As dimensionality increases, visualization becomes difficult, which is why dimensionality reduction techniques such as **PCA** and **LDA** are commonly used.

4️⃣ Eigenvalues & Decomposition


```python
# ===========================================
# Student × Subject Matrix
# ===========================================

student_subject_matrix = np.array([
    [80, 70, 90],
    [60, 85, 75],
    [95, 88, 92]
])

print("Student Subject Matrix:\n")
print(student_subject_matrix)

# ===========================================
# Covariance Matrix
# ===========================================

covariance_matrix = np.cov(student_subject_matrix, rowvar=False)

print("\n========== Covariance Matrix ==========")
print(np.round(covariance_matrix, 2))

# ===========================================
# Eigenvalues & Eigenvectors
# ===========================================

eigenvalues, eigenvectors = np.linalg.eig(covariance_matrix)

print("\n========== Eigenvalues ==========")
print(np.round(eigenvalues, 2))

print("\n========== Eigenvectors ==========")
print(np.round(eigenvectors, 3))

# ===========================================
# LU Decomposition
# ===========================================

P, L, U = lu(student_subject_matrix)

print("\n========== Permutation Matrix (P) ==========")
print(np.round(P, 3))

print("\n========== Lower Triangular Matrix (L) ==========")
print(np.round(L, 3))

print("\n========== Upper Triangular Matrix (U) ==========")
print(np.round(U, 3))

# ===========================================
# Singular Value Decomposition (SVD)
# ===========================================

U_matrix, singular_values, VT_matrix = np.linalg.svd(student_subject_matrix)

print("\n========== U Matrix ==========")
print(np.round(U_matrix, 3))

print("\n========== Singular Values ==========")
print(np.round(singular_values, 3))

print("\n========== Vᵀ Matrix ==========")
print(np.round(VT_matrix, 3))


reduced_dimension = 2

reduced_data = (
    U_matrix[:, :reduced_dimension]
    * singular_values[:reduced_dimension]
)

print("\n========== Reduced Data (2 Dimensions) ==========")
print(np.round(reduced_data, 3))
```

5️⃣ Dimensionality Reduction

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# ===========================================
# Principal Component Analysis (PCA)
# ===========================================


data = pd.DataFrame({
    "Math":     [72, 88, 65, 95, 81],
    "Science":  [75, 90, 68, 97, 84],
    "English":  [70, 85, 60, 92, 79],
    "Physics":  [78, 92, 66, 96, 83],
    "Computer": [80, 94, 70, 98, 86]
},
index=[
    "Student A",
    "Student B",
    "Student C",
    "Student D",
    "Student E"
])

print("========== Original Dataset ==========\n")
print(data)

# ------------------------------------
# Step 1 : Standardize the Data
# ------------------------------------

scaler = StandardScaler()
X_scaled = scaler.fit_transform(data)

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    principal_components,
    columns=[
        "Principal Component 1",
        "Principal Component 2"
    ],
    index=data.index
)

print("\n========== Dataset After PCA ==========\n")
print(pca_df)
print("\n========== Explained Variance Ratio ==========\n")
print(pca.explained_variance_ratio_)

# ------------------------------------
# Visualization
# ------------------------------------

plt.figure(figsize=(8,6))

plt.scatter(
    pca_df["Principal Component 1"],
    pca_df["Principal Component 2"],
    color="blue",
    s=120
)

# Display Student Names

for student, row in pca_df.iterrows():
    plt.text(
        row["Principal Component 1"],
        row["Principal Component 2"],
        student,
        fontsize=10
    )

plt.title("PCA - Student Performance Dataset")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.grid(True)

plt.show()


```

```python
data = pd.DataFrame({
    "Math":     [72, 88, 65, 95, 81],
    "Science":  [75, 90, 68, 97, 84],
    "English":  [70, 85, 60, 92, 79],
    "Physics":  [78, 92, 66, 96, 83],
    "Computer": [80, 94, 70, 98, 86],
    "Result":   ["Pass","Pass","Pass","Fail","Fail"]
})

print("Original Dataset")
print(data)

X = data[["Math","Science","English","Physics","Computer"]]

# Target Labels
y = data["Result"]


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


lda = LinearDiscriminantAnalysis(n_components=1)

X_lda = lda.fit_transform(X_scaled, y)

# Create DataFrame
lda_df = pd.DataFrame({
    "LD1": X_lda.flatten(),
    "Result": y
})

print("\nDataset After LDA")
print(lda_df)

# -------------------------------
# Visualization
# -------------------------------

plt.figure(figsize=(8,2))

colors = ["green" if label=="Pass" else "red"
        for label in y]

plt.scatter(
    lda_df["LD1"],
    np.zeros(len(lda_df)),
    c=colors,
    s=100
)

# Student labels
for i, txt in enumerate(data.index):
    plt.text(
        lda_df["LD1"][i],
        0.02,
        f"S{i+1}"
    )

plt.yticks([])
plt.xlabel("Linear Discriminant 1")
plt.title("LDA - Student Performance Classification")

plt.grid(True)
plt.show()
```
