# 🔹 Part 1: Poisson Distribution

Poisson distribution models the number of events occurring
in a fixed interval of time or space.

Used when:
- Events occur independently
- Average rate (λ) is constant
- Events are discrete counts

📘 Example: Customer Support Calls
Average calls per hour = 5


```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lambda_value = 5

# Probability of exactly 3 calls
prob_3 = poisson.pmf(3, lambda_value)
prob_3
```

```python

x = np.arange(0, 15)
probs = poisson.pmf(x, lambda_value)

plt.plot(x, probs, marker="o")
plt.title("Poisson Distribution (λ = 5)")
plt.xlabel("Number of Calls")
plt.ylabel("Probability")
plt.show()
```

```python
# Real-World Example: Poisson Distribution

# A website receives on average:

# λ = 4

# visitors per minute.

# What is probability of exactly 6 visitors in a minute?

import numpy as np
import math
import matplotlib.pyplot as plt
from scipy.stats import poisson

# -----------------------------
# Step 1: Parameters
# -----------------------------
lam = 4  # average rate
k = 6    # number of events

# -----------------------------
# Step 2: Manual Poisson PMF
# P(X=k) = e^(-λ) * λ^k / k!
# -----------------------------
poisson_manual = (math.exp(-lam) * lam**k) / math.factorial(k)

# -----------------------------
# Step 3: SciPy Poisson PMF
# -----------------------------
poisson_scipy = poisson.pmf(k, lam)

# -----------------------------
# Step 4: CDF (P(X ≤ k))
# -----------------------------
poisson_cdf = poisson.cdf(k, lam)

# PMF => P(X = 6) => stats.poisson.pmf(k, lambda)
# CDF => P(X <= 6) => P(X=1) + P(X=2) + P(X=3) + P(X=4) + P(X=5) + P(X=6) => stats.poisson.cdf(k, lambda)

# -----------------------------
# Step 5: Visualization
# -----------------------------
k_values = np.arange(0, 15)
pmf_values = poisson.pmf(k_values, lam)

plt.bar(k_values, pmf_values)
plt.title("Poisson Distribution (λ=4)")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()

# -----------------------------
# Output
# -----------------------------
print("Manual P(X=6):", poisson_manual)
print("SciPy P(X=6):", poisson_scipy)
print("P(X ≤ 6):", poisson_cdf)
```


# 🔹 Part 2: Z-Score Probability

Z-score formula:

Z = (X - μ) / σ

Used to convert normal variable to standard normal distribution.



📘 Example: Exam Scores
Mean = 70
Std Dev = 10
Find probability score > 85


```python

from scipy.stats import norm

mean = 70
std = 10
x_value = 85

z_score = (x_value - mean) / std
prob_above = 1 - norm.cdf(z_score)

z_score, prob_above
```

```python

x_vals = np.linspace(40, 100, 1000)
pdf_vals = norm.pdf(x_vals, mean, std)

plt.plot(x_vals, pdf_vals)
plt.title("Normal Distribution Curve")
plt.xlabel("Score")
plt.ylabel("Density")
plt.show()
```


# 🔹 Part 3: Probability Density Function (PDF)

PDF shows density at a specific value.
Area under curve = 1.


```python

pdf_value = norm.pdf(85, mean, std)
pdf_value
```


# 🔹 Part 4: Cumulative Distribution Function (CDF)

CDF gives probability that X ≤ x.


```python

cdf_value = norm.cdf(85, mean, std)
cdf_value
```

```python

cdf_vals = norm.cdf(x_vals, mean, std)

plt.plot(x_vals, cdf_vals)
plt.title("Cumulative Distribution Function")
plt.xlabel("Score")
plt.ylabel("Cumulative Probability")
plt.show()
```

```python
# PDF & CDF (Normal Distribution)

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import math

# data = [6.2, 6.0, 5.8, 5.9, 5.4] => PDF
# data = [30, 44, 23, 89, 100, 67] => PMF

# data = 180 => PMF or PDF => ?

# -----------------------------
# Step 1: Given Values
# -----------------------------
mu = 170
sigma = 10
x = 180

# -----------------------------
# Step 2: Manual PDF
# -----------------------------
pdf_manual = (1 / (sigma * math.sqrt(2 * math.pi))) * \
             math.exp(-((x - mu)**2) / (2 * sigma**2))

# -----------------------------
# Step 3: SciPy PDF & CDF
# -----------------------------
pdf_scipy = norm.pdf(x, mu, sigma)
cdf_value = norm.cdf(x, mu, sigma)

# -----------------------------
# Step 4: Visualization
# -----------------------------
x_values = np.linspace(mu - 3*sigma, mu + 3*sigma, 500)
pdf_values = norm.pdf(x_values, mu, sigma)
cdf_values = norm.cdf(x_values, mu, sigma)

plt.figure(figsize=(12,5))

# PDF Plot
plt.subplot(1,2,1)
plt.plot(x_values, pdf_values)
plt.axvline(x, color="red", linestyle="--")
plt.title("Probability Density Function (PDF)")
plt.xlabel("Height")
plt.ylabel("Density")

# CDF Plot
plt.subplot(1,2,2)
plt.plot(x_values, cdf_values)
plt.axvline(x, color="red", linestyle="--")
plt.axhline(cdf_value, color="red", linestyle="--")
plt.title("Cumulative Distribution Function (CDF)")
plt.xlabel("Height")
plt.ylabel("Probability")

plt.show()

# -----------------------------
# Output
# -----------------------------
print("Manual PDF at 180:", pdf_manual)
print("SciPy PDF at 180:", pdf_scipy)
print("P(X ≤ 180):", cdf_value)
```
