# 🔹 Part 1: What is Statistical Distribution?

A statistical distribution describes how values of a variable are spread or distributed.

It shows:
- Possible values
- Probability of each value

Two main types:

1. **Discrete Distribution** (e.g., Binomial)
2. **Continuous Distribution** (e.g., Normal)


```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

delivery_time = np.random.normal(loc=30, scale=5, size=200)

plt.hist(delivery_time, bins=20)
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.show()
```

##### **Binomial Distribution (Discrete)**

image.png

```python

import numpy as np
import math
from scipy.stats import binom, norm

# ===============================
# BINOMIAL DISTRIBUTION
# ===============================

# Example: 10 trials, probability of success = 0.5
n = 10
p = 0.5
k = 4  # number of successes

# ----- Manual Binomial -----
comb = math.comb(n, k)
binomial_manual = comb * (p**k) * ((1-p)**(n-k))

# ----- SciPy Binomial -----
binomial_scipy = binom.pmf(k, n, p)

# ===============================
# OUTPUT
# ===============================

print("BINOMIAL DISTRIBUTION")
print("---------------------")
print("Manual P(X=4):", binomial_manual)
print("SciPy P(X=4):", binomial_scipy)
```

##### **Normal Distribution (Continuous)**

image.png

```python
import numpy as np
import math
from scipy.stats import binom, norm

# ===============================
# NORMAL DISTRIBUTION
# ===============================

mu = 50
sigma = 10
x = 60

# ----- Manual Normal PDF -----
normal_manual = (1 / (sigma * math.sqrt(2 * math.pi))) * \
                math.exp(-((x - mu)**2) / (2 * sigma**2))

# ----- SciPy Normal PDF -----
normal_scipy = norm.pdf(x, mu, sigma)

# ----- CDF (Probability P(X <= x)) -----
normal_cdf = norm.cdf(x, mu, sigma)

# ===============================
# OUTPUT
# ===============================

print("\nNORMAL DISTRIBUTION")
print("-------------------")
print("Manual PDF at x=60:", normal_manual)
print("SciPy PDF at x=60:", normal_scipy)
print("SciPy CDF P(X<=60):", normal_cdf)
```


# 🔹 Part 2: Discrete vs Continuous Distribution


```python

# Discrete example
defects = np.random.binomial(n=10, p=0.2, size=100)

plt.hist(defects, bins=10)
plt.title("Discrete Distribution - Defects")
plt.xlabel("Number of Defects")
plt.ylabel("Frequency")
plt.show()
```


# 🔹 Part 3: Q-Q Plot

Used to check whether data follows normal distribution.

A Q–Q Plot compares the quantiles of your dataset with the quantiles of a theoretical distribution (usually Normal).
image.png


```python

import scipy.stats as stats

stats.probplot(delivery_time, dist="norm", plot=plt)
plt.title("Q-Q Plot for Delivery Time")
plt.show()
```


# 🔹 Part 4: Bernoulli Distribution

Represents a single trial with two outcomes (0 or 1).

A Bernoulli distribution models a single trial experiment with:|
- Only two outcomes
- Success (1)
- Failure (0)
- Probability of success = p
- Probability of failure = 1 − p

# 🔹 Part 5: Binomial Distribution

Number of successes in n independent Bernoulli trials.

A Binomial distribution models:
- n independent Bernoulli trials
- Probability of exactly k successes


image.png

```python
# Bernoulli & Binomial Distribution
# Manual + SciPy (Single Block)

import numpy as np
import math
from scipy.stats import bernoulli, binom

# ===============================
# PART 1: BERNOULLI DISTRIBUTION
# ===============================

p = 0.6  # probability of success

# Single trial outcomes
x_success = 1
x_failure = 0

# ----- Manual Bernoulli -----
bern_success_manual = p**x_success * (1-p)**(1-x_success)
bern_failure_manual = p**x_failure * (1-p)**(1-x_failure)

# ----- SciPy Bernoulli -----
bern_success_scipy = bernoulli.pmf(1, p)
bern_failure_scipy = bernoulli.pmf(0, p)

# ===============================
# PART 2: BINOMIAL DISTRIBUTION
# ===============================

n = 5   # number of trials
k = 3   # number of successes

# ----- Manual Binomial -----
comb = math.comb(n, k)
binomial_manual = comb * (p**k) * ((1-p)**(n-k))

# ----- SciPy Binomial -----
binomial_scipy = binom.pmf(k, n, p)

# Cumulative probability P(X ≤ 3)
binomial_cdf = binom.cdf(k, n, p)

# ===============================
# OUTPUT
# ===============================

print("BERNOULLI DISTRIBUTION")
print("----------------------")
print("Manual P(Success=1):", bern_success_manual)
print("SciPy P(Success=1):", bern_success_scipy)
print("Manual P(Failure=0):", bern_failure_manual)
print("SciPy P(Failure=0):", bern_failure_scipy)

print("\nBINOMIAL DISTRIBUTION")
print("---------------------")
print("Manual P(X=3):", binomial_manual)
print("SciPy P(X=3):", binomial_scipy)
print("SciPy CDF P(X<=3):", binomial_cdf)
```

# 📌 Mini Assignment

A website has conversion probability = 0.25.

Tasks:
1. Model one visitor using Bernoulli distribution.
2. Model 15 visitors using Binomial distribution.
3. Plot binomial probability distribution.
4. Find probability exactly 5 users convert.
5. Simulate 1000 experiments and compare empirical vs theoretical probability.
6. Use Q-Q plot to check sampling distribution of sample means.


```python
# A website has conversion probability = 0.25.

# Tasks:
# 1. Model one visitor using Bernoulli distribution.
# 2. Model 15 visitors using Binomial distribution.
# 3. Plot binomial probability distribution.
# 4. Find probability exactly 5 users convert.
# 5. Simulate 1000 experiments and compare empirical vs theoretical probability.
# 6. Use Q-Q plot to check sampling distribution of sample means.
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, norm
import scipy.stats as stats

# Parameters
p = 0.25  # conversion probability
n_visitors = 15
n_experiments = 1000

# 1. Bernoulli distribution for one visitor
bernoulli_dist = bernoulli(p)
print("Bernoulli P(Conversion=1):", bernoulli_dist.pmf(1))
print("Bernoulli P(Conversion=0):", bernoulli_dist.pmf(0))
# 2. Binomial distribution for 15 visitors
binomial_dist = binom(n_visitors, p)
# 3. Plot binomial probability distribution
k_values = np.arange(0, n_visitors + 1)
binomial_probs = binomial_dist.pmf(k_values)
plt.bar(k_values, binomial_probs)
plt.title("Binomial Distribution - Number of Conversions")
plt.xlabel("Number of Conversions")
plt.ylabel("Probability")
plt.xticks(k_values)
plt.show()
# 4. Probability exactly 5 users convert
prob_5_conversions = binomial_dist.pmf(5)
print("P(Exactly 5 Conversions):", prob_5_conversions)
# 5. Simulate 1000 experiments
simulated_conversions = np.random.binomial(n_visitors, p, n_experiments)
empirical_prob_5 = np.mean(simulated_conversions == 5)
print("Empirical P(Exactly 5 Conversions):", empirical_prob_5)
# 6. Q-Q plot for sample means
sample_means = [np.mean(np.random.binomial(n_visitors, p, n_visitors)) for _ in range(n_experiments)]
norm_means = norm.rvs(loc=n_visitors*p, scale=np.sqrt(n_visitors*p*(1-p)), size=n_experiments)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.title("Q-Q Plot for Sample Means")
stats.probplot(sample_means, dist="norm", plot=plt)
plt.subplot(1, 2, 2)
plt.title("Q-Q Plot for Theoretical Normal")
stats.probplot(norm_means, dist="norm", plot=plt)
plt.tight_layout()
plt.show()
```
