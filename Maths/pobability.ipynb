```python
import pandas as pd
import numpy as np

order_df = pd.DataFrame({
    "Day":range(1,31),
    "orders":np.random.randint(60,160,size=30)
})
order_df.head()
```

```python
high_orders_days = order_df[order_df["orders"] > 120]
p_orders_gt_120 = len(high_orders_days) / len(order_df)
print(f"Probability of orders greater than 120: {p_orders_gt_120:.2f}")
print(f"Probability in %: {p_orders_gt_120 * 100:.2f}%%")
```

```python
data = {
    'index':[0,1,2,3,4],
    'Costomer ID':[101,102,103,104,105],
    'gender':['M','F','M','F','M'],
    'Age':[25,30,22,28,35],
    'purchased':[1,0,1,1,1]
}
df = pd.DataFrame(data)
print(df)
```

## Simle Event

```python
p_female = len(df[df["gender"] == "F"]) / len(df)
print(f"Probability of a customer being female: {p_female:.2f}")
```

## Compund Event

```python
female_and_purchased = df[(df["gender"] == "F") & (df["purchased"] == 1)]
p_compund = len(female_and_purchased) / len(df)
print(f"Probability of a customer being female and having purchased: {p_compund:.2f}")  
```

```python
p_female = len(df[df["gender"] == "F"]) / len(df)
p_purchased = len(df[df["purchased"] == 1]) / len(df)
female_and_purchased = df[(df["gender"] == "F") & (df["purchased"] == 1)]
p_joint = len(female_and_purchased) / len(df)

print(f"Probability of a customer being female: {p_female:.2f}")
print(f"Probability of a customer having purchased: {p_purchased:.2f}")
print(f"Probability of a customer being female and having purchased: {p_joint:.2f}")
print(f"p(female)*p(purchased): {(p_female * p_purchased):.2f}")

if p_joint == p_female * p_purchased:
    print("The events are independent.")
else:
    print("The events are not independent.")
```
