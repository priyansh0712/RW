```python
point1 = np.array([2, 3])
point2 = np.array([5, 7])
distance = np.sqrt((point2[0] - point1[0])**2 + (point2[1] - point1[1])**2)
print("Euclidean Distance between point1 and point2:", distance)
```

```python
points = np.random.rand(5, 2) * 100
distances = []
for i in range(len(points)):
    for j in range(i + 1, len(points)):
        dist = np.linalg.norm(points[i] - points[j])
        distances.append((i, j, dist))
print("Random Points:\n", points)
print("Distances between points:")
for i, j, dist in distances:
    print(f"Distance between point {i} and point {j}: {dist:.2f}")
```

```python
from scipy.spatial.distance import euclidean
pointA = np.array([2, 3])
pointB = np.array([7,8])

distance_manual = np.sqrt((pointB[0] - pointA[0])**2 + (pointB[1] - pointA[1])**2)
print("Distance is(manual method):", distance_manual)

distance_scipy = euclidean(pointA, pointB)
print("Distance is(scipy method):", distance_scipy)
```

```python
from scipy.spatial.distance import cdist
points = np.random.rand(10, 2) * 100
distance_matrix = cdist(points, points)
print("Random Points:\n", points)
print("Pairwise Distance Matrix:\n", distance_matrix)
```

```python
v = np.array([4,-3,7])
print("L1(Mahhatam distance):",np.sum(np.abs(v)))
print("L1(Mahhatam distance):", np.linalg.norm(v, ord=1))
print("L2(Euclidean distance):", np.linalg.norm(v, ord=2))
print("L2(Euclidean distance):", np.sqrt(np.sum(v**2)))
```

```python
def convert_to_unit_vector(vector):
    magnitude = np.linalg.norm(vector)
    unit_vector = vector / magnitude
    return unit_vector
print("Unit vector of Point5D:", convert_to_unit_vector([1, 2, 3, 4, 5]))
```

```python
vectors = np.random.rand(5, 3)
unit_vectors = [convert_to_unit_vector(v) for v in vectors]
print("Unit vectors:")
for i, unit_vector in enumerate(unit_vectors):
    print(f"Vector {i+1}: {unit_vector}")
    print(f"Length of Vector {i+1}: {np.linalg.norm(unit_vector)}")
    
```
