# 🔹 Part 2: Vector vs Point

## Point
Represents a location in space.
Example: (3, 4)

## Vector
Represents direction and magnitude.
Example: [3, 4]


```python

import numpy as np
import matplotlib.pyplot as plt

point = np.array([3, 4])
vector = np.array([3, 4])

plt.scatter(point[0], point[1])
plt.quiver(0, 0, vector[0], vector[1], angles="xy", scale_units="xy", scale=1)
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.title("Vector and Point Representation")
plt.show()
```


# 🔹 Part 3: Real-World Example – Student Scores as Vectors

Student A = [80, 75]  
Student B = [65, 90]


```python

student_A = np.array([80, 75])
student_B = np.array([65, 90])

student_A, student_B
```


# 🔹 Part 4: Vector Operations


## ➤ Vector Addition

```python

addition = student_A + student_B
addition
```

## ➤ Vector Subtraction

```python

subtraction = student_A - student_B
subtraction
```

## ➤ Dot Product

```python

dot_product = np.dot(student_A, student_B)
dot_product
```

## ➤ Magnitude of Vector

image.png

```python

manual_magnitude_A = (student_A[0] ** 2 + student_A[1] ** 2) ** 0.5
manual_magnitude_B = (student_B[0] ** 2 + student_B[1] ** 2) ** 0.5

magnitude_A = np.linalg.norm(student_A)
magnitude_B = np.linalg.norm(student_B)

print(manual_magnitude_A, manual_magnitude_B)
print(magnitude_A, magnitude_B)
```

## ➤ Angle Between Two Vectors

image.png

```python
import math

cos_theta = dot_product / (magnitude_A * magnitude_B)
angle_radian = np.arccos(cos_theta)  # same as math.acos(cos_theta)

print(angle_radian)

angle_degree = np.degrees(angle_radian)  # same as math.degrees(angle_radian)

print(angle_degree)
```


# 🔹 Visualization of Student Vectors


```python

plt.quiver(0, 0, student_A[0], student_A[1], angles='xy', scale_units='xy', scale=1, color='red', label='Student A')
plt.quiver(0, 0, student_B[0], student_B[1], angles='xy', scale_units='xy', scale=1, color='blue', label='Student B')

plt.xlim(0, 100)
plt.ylim(0, 100)

plt.grid(True)
plt.title("Student Performance Vectors")
plt.xlabel("Experience (years)")
plt.ylabel("Performance score")
plt.legend()
plt.show()
```

```python
# Vector & Its Operations

import numpy as np
import math

# Define vectors
a = np.array([2, 3])
b = np.array([4, 1])

# -----------------------------
# 1. Vector Addition
# -----------------------------
addition = a + b

# -----------------------------
# 2. Vector Subtraction
# -----------------------------
subtraction = a - b

# -----------------------------
# 3. Scalar Multiplication
# -----------------------------
scalar = 3
scalar_multiplication = scalar * a

# -----------------------------
# 4. Magnitude (Length)
# -----------------------------
magnitude_a = np.linalg.norm(a)
magnitude_b = np.linalg.norm(b)

# -----------------------------
# 5. Dot Product
# -----------------------------
dot_product = np.dot(a, b)

# Angle between vectors
angle = math.degrees(
    math.acos(dot_product / (magnitude_a * magnitude_b))
)

# -----------------------------
# 6. Cross Product (3D Example)
# -----------------------------
a3 = np.array([1, 2, 3])
b3 = np.array([4, 5, 6])
cross_product = np.cross(a3, b3)

# -----------------------------
# Output
# -----------------------------
print("Vector a:", a)
print("Vector b:", b)

print("\nAddition:", addition)
print("Subtraction:", subtraction)
print("Scalar Multiplication (3a):", scalar_multiplication)

print("\nMagnitude of a:", magnitude_a)
print("Magnitude of b:", magnitude_b)

print("\nDot Product:", dot_product)
print("Angle between a and b (degrees):", angle)

print("\nCross Product (3D example):", cross_product)
```
